##**Задачи - JOIN (Оба уровня)**

В этом тесте вам предстоит решить практические задачи на тему "JOIN".

Найдите все бронирования, которые получили оценку 5, и выведите их ID, даты заезда и выезда и значение оценки. Названия колонок оставьте оригинальными. Отсортируйте результаты по ID бронирования и выведите первые 50 записей.

```python
SELECT 
    b.booking_id,
    b.check_in_date,
    b.check_out_date,
    r.rating_value
FROM 
    bookings b
JOIN 
    ratings r ON b.booking_id = r.booking_id
WHERE 
    r.rating_value = 5
ORDER BY 
    b.booking_id
LIMIT 50;
```

Найдите бронирования, которые были сделаны клиентами, у которых телефон начинается с "8910", и укажите ID бронирования, дату заезда, дату выезда, имя, фамилию и номер телефона клиента.

```python
SELECT 
    b.booking_id,
    b.check_in_date,
    b.check_out_date,
    c.first_name,
    c.last_name,
    c.phone_number
FROM 
    bookings b
JOIN 
    clients c ON b.renter_id = c.id
WHERE 
    c.phone_number LIKE '8910%'
ORDER BY 
    b.booking_id;
```

```python
SELECT 
    b.booking_id,
    b.check_in_date,
    b.check_out_date,
    c.first_name,
    c.last_name
FROM 
    bookings b
JOIN 
    clients c ON b.renter_id = c.id
ORDER BY 
    b.booking_id
LIMIT 50;
```

Выведите id (поле так и назовите - id), фамилию, имя и номер телефона клиента, а также номер комнаты для бронирований, где оценка за бронирование ниже 4. Колонки назовите так, как они хранятся в таблицах. Отсортируйте результаты по ID клиента и выведите первые 50 записей.

```python
SELECT 
    c.id AS id,
    c.last_name,
    c.first_name,
    c.phone_number,
    b.room_number
FROM 
    bookings b
JOIN 
    clients c ON b.renter_id = c.id
JOIN 
    ratings r ON b.booking_id = r.booking_id
WHERE 
    r.rating_value < 4
ORDER BY 
    c.id
LIMIT 50;
```

Найдите ID, имена и фамилии клиентов, которые забронировали номера типа "Люкс" и проживают в городе "Рязань", и укажите ID бронирования, даты заезда и выезда. Отсортируйте результаты по ID клиента, а затем по ID бронирования.

```python
SELECT 
    c.id,
    c.first_name,
    c.last_name,
    b.booking_id,
    b.check_in_date,
    b.check_out_date
FROM 
    bookings b
JOIN 
    rooms r ON b.room_number = r.room_number
JOIN 
    clients c ON b.renter_id = c.id
WHERE 
    r.type_name = 'Люкс' 
    AND c.address LIKE '%Рязань%'
ORDER BY 
    c.id, b.booking_id;
```

Выведите список номеров комнат, бронь которых полностью укладывается в период с 1 марта 2022 года по 15 марта 2022 года (включительно - т.е. и старт, и конец брони должны быть внутри этого периода), при условии, что суточная цена номера была выше 20000 и номер был оборудован минибаром.. Укажите также id, имя и фамилию клиента, забронировавшего номер.  Отсортируйте результаты по номеру комнаты, и ID клиента.

```python
SELECT 
    r.room_number,
    c.id,
    c.first_name,
    c.last_name
FROM 
    bookings b
JOIN 
    rooms r ON b.room_number = r.room_number
JOIN 
    clients c ON b.renter_id = c.id
WHERE 
    b.check_in_date >= '2022-03-01'
    AND b.check_out_date <= '2022-03-16'
    AND r.price_per_night > 20000
    AND r.has_minibar = 1
GROUP BY
    r.room_number, c.id
ORDER BY 
    r.room_number, c.id;
```

Найдите все бронирования, которые были оценены на '5', и укажите ID бронирования, дату заезда и тип забронированного номера. Отсортируйте результаты по ID бронирования и выведите первые 50 записей.

```python
SELECT 
    b.booking_id,
    b.check_in_date,
    r.type_name
FROM 
    bookings b
JOIN 
    ratings rts ON b.booking_id = rts.booking_id
JOIN 
    rooms r ON b.room_number = r.room_number
WHERE 
    rts.rating_value = 5
ORDER BY 
    b.booking_id
LIMIT 50;
```

Вывести список клиентов (ID, имя и фамилию), которые забронировали номера с мини-баром и оплатили бронирование 'картой', с указанием ID бронирования и ID платежа, отсортированный по ID клиента, ID бронирования и ID платежа. Ограничить вывод первыми 15 записями.

```python
SELECT 
    c.id,
    c.first_name,
    c.last_name,
    b.booking_id,
    p.payment_id
FROM 
    clients c
JOIN 
    bookings b ON c.id = b.renter_id
JOIN 
    rooms r ON b.room_number = r.room_number
JOIN 
    payments p ON b.booking_id = p.booking_id
WHERE 
    r.has_minibar = 1
    AND p.payment_method = 'картой'
ORDER BY 
    c.id, b.booking_id, p.payment_id
LIMIT 15;
```

Найдите бронирования, которые были сделаны клиентами с адресом в городе "Москва", и укажите номер бронирования, дату заезда, дату выезда, тип номера и сумму платежа. Отсортируйте результаты по сумме платежа и ID бронирования. Ограничить вывод первыми 20 записями.

```python
SELECT 
    b.booking_id,
    b.check_in_date,
    b.check_out_date,
    r.type_name,
    p.amount_paid
FROM 
    clients c
JOIN 
    bookings b ON c.id = b.renter_id
JOIN 
    rooms r ON b.room_number = r.room_number
JOIN 
    payments p ON b.booking_id = p.booking_id
WHERE 
    c.address LIKE '%Москва%'
ORDER BY 
    p.amount_paid, b.booking_id
LIMIT 20;
```
