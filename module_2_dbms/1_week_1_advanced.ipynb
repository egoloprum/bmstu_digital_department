В этом тесте вам предстоит решить практические задачи продвинутого уровня на тему "Простые запросы".

Вывести наименьшую сумму платежа, дату этого платежа и метод оплаты.

```python
SELECT amount_paid, payment_date, payment_method
FROM payments
ORDER BY amount_paid ASC
LIMIT 1;
```

Вывести вторую по величине сумму платежа, дату этого платежа и метод оплаты.

```python
SELECT amount_paid, payment_date, payment_method
FROM payments
WHERE amount_paid = (
    SELECT DISTINCT amount_paid
    FROM payments
    ORDER BY amount_paid DESC
    LIMIT 1 OFFSET 1
)
limit 1
;
```

Вывести всю информацию о бронировании с самой поздней датой заезда.

```python
SELECT *
FROM bookings
WHERE check_in_date = (
    SELECT MAX(check_in_date)
    FROM bookings
);
```

Определить, во сколько обойдется недельное пребывание в комнате. Для каждой комнаты вывести ее номер, цену за ночь и колонку week_price, в которой будет указана цена за неделю пребывания.

```python
SELECT 
    room_number, 
    price_per_night, 
    price_per_night * 7 AS week_price
FROM 
    rooms;
```

Определить цену за квадратный метр для каждой комнаты. Для каждой комнаты вывести ее номер, цену за ночь, площадь и колонку sqm_price, в которой будет указана стоимость одного квадратного метра.

```python
SELECT 
    room_number, 
    price_per_night, 
    room_size_sqm, 
    CAST(price_per_night / room_size_sqm AS INT) AS sqm_price
FROM 
    rooms;
```
