##**Задачи - Фильтрация данных (Оба уровня)**

В этом тесте вам предстоит решить практические задачи на тему "Фильтрация данных".

Получить всю информацию о комнатах с минибарами.

```python
SELECT *
FROM rooms
WHERE has_minibar = 1;
```

Получить всю информацию о платежах до 1500 р. включительно. Ограничьте вывод 20 записями.

```python
SELECT *
FROM payments
WHERE amount_paid <= 1500
LIMIT 20;
```

Найти все комнаты с минибаром и ценой за ночь менее 10 тыс.

```python
SELECT *
FROM rooms
WHERE has_minibar = 1 AND price_per_night < 10000;
```

Вывести список всех номеров категории 'Люкс' с площадью более 45 м2.

```python
SELECT *
FROM rooms
WHERE type_name = 'Люкс' AND room_size_sqm > 45;
```

Определить сколько гостей остались довольными отдыхом. Вывести идентификатор оценки и значение оценок для бронирований с оценкой не ниже 4. Ограничьте вывод 50 записями.

```python
SELECT rating_id, rating_value
FROM ratings
WHERE rating_value >= 4
LIMIT 50;
```

Получить всю информацию о клиентах, которые проживают в городе Нижнекамск (имеется в виду именно г. Нижнекамск, не село или что-то иное).

```python
SELECT *
FROM clients
WHERE address LIKE 'г. Нижнекамск%';
```

Вывести информацию о первых 60 платежах, которые были проведены картой.

```python
SELECT *
FROM payments
WHERE payment_method = 'картой'
LIMIT 60;
```

Вывести номера комнат и их типы, где есть минибар и вместимость больше трех человек.

```python
SELECT room_number, type_name
FROM rooms
WHERE has_minibar = 1 AND max_occupancy > 3;
```

Вывести всю информацию о самой большой комнате без минибара.

```python
SELECT *
FROM rooms
WHERE room_size_sqm = (
    SELECT MAX(room_size_sqm)
    FROM rooms
    WHERE has_minibar = 0
);
```

Вывести всю информацию о самом большом платеже с методом оплаты 'СБП'. Колонки должны быть выведены в том порядке, в котором они изначально хранятся в таблице.

```python
SELECT *
FROM payments
WHERE amount_paid = (
    SELECT MAX(amount_paid)
    FROM payments
    WHERE payment_method = 'СБП'
);
```

Вывести всю информацию по всем платежам с суммой платежа не менее 50 тыс., отсортировав по убыванию суммы платежа. Ограничьте вывод 40 записями.

```python
SELECT *
FROM payments
WHERE amount_paid >= 50000
ORDER BY amount_paid DESC
LIMIT 40;
```

Вывести имена и фамилии клиентов, которые не указали электронную почту 

```python
SELECT first_name, last_name
FROM clients
WHERE email IS NULL OR email = '';
```
