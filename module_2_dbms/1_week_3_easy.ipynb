##**Задачи - Группировка и агрегация данных (Оба уровня)**

В этом тесте вам предстоит решить практические задачи на тему "Группировка и агрегация данных".

Сколько всего бронирований было сделано? (total_bookings)

```python
SELECT 
    COUNT(*) AS total_bookings
FROM 
    bookings;
```

Какова средняя цена за ночь (avg_price_per_night) для всех номеров?

```python
SELECT 
    AVG(price_per_night) AS avg_price_per_night
FROM 
    rooms;
```

Какова максимальная вместимость (max_room_occupancy) номера в отеле?

```python
SELECT 
    MAX(max_occupancy  ) AS max_room_occupancy
FROM 
    rooms;
```

Какова общая сумма всех оплат (total_amount_paid), совершенных клиентами?

```python
SELECT 
    SUM(amount_paid) AS total_amount_paid
FROM 
    payments;
```

Какова минимальная цена за ночь среди номеров с мини-баром? (min_price_with_minibar)

```python
SELECT 
    MIN(price_per_night) AS min_price_with_minibar
FROM 
    rooms
WHERE 
    has_minibar = 1;  -- Assuming 1 indicates the presence of a mini-bar
```

Сколько номеров имеют размер больше 40 кв. метров? (num_large_rooms)

```python
SELECT 
    COUNT(*) AS num_large_rooms
FROM 
    rooms
WHERE 
    room_size_sqm > 40;  -- Filtering for rooms larger than 40 square meters
```

Сколько бронирований было оплачено методом 'СБП'? (bookings_paid_sbp)

```python
SELECT 
    COUNT(*) AS bookings_paid_sbp
FROM 
    payments
WHERE 
    payment_method = 'СБП';  -- Filtering for payments made using 'СБП'
```

Выведите количество номеров для каждого типа номера. (В результате должно быть 2 колонки: type_name и колонка с подсчетом количества number_of_rooms.)

```python
SELECT 
    type_name,
    COUNT(*) AS number_of_rooms
FROM 
    rooms
GROUP BY 
    type_name;
```

Выведите среднюю сумму оплаты для каждого способа оплаты. (avg_payment_amount)

```python
SELECT 
    payment_method,
    AVG(amount_paid) AS avg_payment_amount
FROM 
    payments
GROUP BY 
    payment_method;
```

Для каждого типа номера в отеле вывести общее количество доступных номеров(total_rooms)

```python
SELECT 
    type_name,
    COUNT(*) AS total_rooms
FROM 
    rooms
GROUP BY 
    type_name;
```

Для первых 60 клиентов вывести их идентификатор и количество сделанных ими бронирований (booking_count)

```python
SELECT 
    c.id as renter_id,
    COUNT(b.booking_id) AS booking_count
FROM 
    clients c
LEFT JOIN 
    bookings b ON c.id = b.renter_id
GROUP BY 
    c.id
ORDER BY 
    c.id  -- Adjust this if you want a different ordering
LIMIT 60;
```
