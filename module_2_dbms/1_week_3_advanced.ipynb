Определите первых 60 клиентов (renter_id), которые сделали более 4 бронирований, с указанием количества их бронирований (booking_count).

```python
SELECT renter_id, COUNT(booking_id) AS booking_count
FROM bookings
GROUP BY renter_id
HAVING COUNT(booking_id) > 4
LIMIT 60;
```

Выведите количество номеров для каждого типа номеров (num_rooms_with_minibar), где количество номеров с мини-баром превышает 2.

```python
SELECT type_name, COUNT(*) AS num_rooms_with_minibar
FROM rooms
WHERE has_minibar = 1
GROUP BY type_name
HAVING COUNT(*) > 2;
```

Выведите топ-3 клиента (renter_id), которые сделали наибольшее количество бронирований (booking_count). Если максимальное количество бронирований сделало больше трех клиентов, выведите среди них первых троих по возрастанию renter_id.

```python
WITH BookingCounts AS (
    SELECT renter_id, COUNT(booking_id) AS booking_count
    FROM bookings
    GROUP BY renter_id
),
MaxCount AS (
    SELECT MAX(booking_count) AS max_booking_count
    FROM BookingCounts
)
SELECT renter_id, booking_count
FROM BookingCounts
WHERE booking_count = (SELECT max_booking_count FROM MaxCount)
ORDER BY renter_id
LIMIT 3;
```

Определите количество бронирований (booking_count) и общую сумму оплат (total_amount_paid) для каждого способа оплаты.

```python
SELECT payment_method, 
       COUNT(booking_id) AS booking_count, 
       SUM(amount_paid) AS total_amount_paid
FROM payments
GROUP BY payment_method;
```

Для каждого типа номера определите минимальную (min_price), максимальную (max_price) и среднюю (avg_price) цену за ночь.

```python
SELECT type_name, 
       MIN(price_per_night) AS min_price, 
       MAX(price_per_night) AS max_price, 
       AVG(price_per_night) AS avg_price
FROM rooms
GROUP BY type_name;
```

Сколько клиентов не оставили свой номер телефона в базе данных? (clients_without_numbers)

```python
SELECT COUNT(*) AS clients_without_numbers
FROM clients
WHERE phone_number IS NULL OR phone_number = '';
```

Вывести количество бронирований для каждого номера за 2020 год. Колонку с количеством назовите booking_count_2020. (Учитывать только дату заезда (check_in_date) - только она обязательно должна быть внутри 2020 года.)
 
Примечание: В sqlite и, соответственно, в ЛМС недоступны функции postgres для работы с датой. Для извлечения года из даты используйте strftime('%Y', date) либо воспользуйтесь более простыми способами вроде неравенств.

```python
SELECT room_number  , 
       COUNT(*) AS booking_count_2020
FROM bookings
WHERE check_in_date >= '2020-01-01' AND check_in_date < '2021-01-01'
GROUP BY room_number  ;
```

Определите номера, которые были забронированы более 18 раз с января по март 2020 года, с указанием количества бронирований (num_bookings) для каждого номера
 

Примечание: В sqlite и, соответственно, в ЛМС недоступны функции postgres для работы с датой. Для извлечения года из даты используйте strftime('%Y', date). Для извлечения номера месяца используйте CAST(strftime('%m', date) AS INTEGER). Либо воспользуйтесь более простыми способами вроде неравенств.

```python
SELECT room_number  , 
       COUNT(*) AS num_bookings
FROM bookings
WHERE check_in_date >= '2020-01-01' AND check_in_date < '2020-04-01'
GROUP BY room_number  
HAVING COUNT(*) > 18;
```
