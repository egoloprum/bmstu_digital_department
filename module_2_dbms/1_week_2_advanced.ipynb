##**Задачи - Фильтрация данных (Продвинутый уровень)**

В этом тесте вам предстоит решить практические задачи продвинутого уровня на тему "Фильтрация данных".

Вывести список номеров всех комнат, которые были заняты хотя бы в течение одного дня в марте 2021 года. Значения повторяться не должны. Отсортируйте номера по возрастанию.

```python
SELECT DISTINCT 
    r.room_number
FROM 
    rooms r
JOIN 
    bookings b ON r.room_number = b.room_number
WHERE 
    b.check_in_date >= '2021-03-01' AND b.check_out_date < '2021-04-01'  -- Filter for March 2021
ORDER BY 
    r.room_number;  -- Sort by room number in ascending order
```

Получить всю информацию о первых 60 клиентах, чьи номера телефонов начинаются с '+7'

```python
SELECT *
FROM clients
WHERE phone_number LIKE '+7%'
LIMIT 60;
```

Получить всю информацию о бронированиях комнаты №15 и датой заезда после 2020-02-15, отсортировать по увеличению даты заезда. Ограньчьте вывод 50 записями.

```python
SELECT *
FROM bookings
WHERE room_number = 15 AND check_in_date > '2020-02-15'  -- Filter for room number 15 and check-in date
ORDER BY check_in_date ASC  -- Sort by check-in date in ascending order
LIMIT 50;  -- Limit the output to 50 records
```

Вывести имена и фамилии клиентов, которые не указали ни номер телефона, ни почту.

```python
SELECT 
    first_name, 
    last_name
FROM 
    clients
WHERE 
    (phone_number IS NULL OR phone_number = '') AND 
    (email IS NULL OR email = '');
```

Вывести всю информацию об оплатах, которые были произведены наличными, где сумма платежа составила от 10000 до 25000, отсортировать в порядке уменьшения суммы оплаты. Ограничьте вывод 50 записями.

```python
SELECT *
FROM payments
WHERE payment_method = 'наличные'  -- Assuming 'наличные' indicates cash payments
  AND amount_paid BETWEEN 10000 AND 25000  -- Filter for payment amounts between 10,000 and 25,000
ORDER BY amount_paid DESC  -- Sort by payment amount in descending order
LIMIT 50;  -- Limit the output to 50 records
```

Для всех комнат вывести номер, площадь в кв. метрах и дополнительную колонку 'room_size', в которой будет указан один из трех показателей (маленькая/средняя/большая). В случае если площадь комнаты менее 35 кв. м., то данная комната считается маленькой, комната с площадью от 35 до 50 кв. м. считается средней и комната более 50 кв. м. считается большой.

```python
SELECT 
    room_number,
    room_size_sqm,
    CASE 
        WHEN room_size_sqm < 35 THEN 'маленькая'
        WHEN room_size_sqm BETWEEN 35 AND 50 THEN 'средняя'
        ELSE 'большая'
    END AS room_size
FROM 
    rooms;
```

Для первых 60 клиентов (доп. условие по  id<=60 учитывать не нужно) вывести имя, фамилию, эл. почту, номер телефона и дополнительную колонку 'сведения', в которой будет указан один из трех показателей (полные/частичные/отсутствуют). В случае если клиент указал номер телефона и эл. почту, то его данные считаются полными, если указан только номер или эл. почта - частичные, в противном случае - отсутствуют.

```python
SELECT 
    first_name,
    last_name,
    email,
    phone_number,
    CASE 
        WHEN phone_number IS NOT NULL AND email IS NOT NULL THEN 'полные'  -- Both phone and email are provided
        WHEN phone_number IS NOT NULL OR email IS NOT NULL THEN 'частичные'  -- Either phone or email is provided
        ELSE 'отсутствуют'  -- Neither phone nor email is provided
    END AS сведения
FROM 
    clients
LIMIT 60;  -- Limit the output to the first 60 clients
```

Вывести всю информацию о бронированиях комнат под номерами 1, 5, 20 за 2018 год (фильровать только по началу бронирования). Ограничьте вывод 60 записями

Примечание: в sqlite нет многих функций, которые есть в postgres. Например, в ЛМС не получится использовать EXTRACT, date_trunc и др. Но отфильтровать данные можно и более простым способом.

```python
SELECT *
FROM bookings
WHERE room_number IN (1, 5, 20)  -- Filter for specific room numbers
  AND check_in_date LIKE '2018%'  -- Filter for check-in dates in the year 2018
LIMIT 60;  -- Limit the output to 60 records
```
