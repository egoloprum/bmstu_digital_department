Вывести номер каждой комнаты, ее тип и площадь.

```python
SELECT room_number, type_name   , room_size_sqm
FROM rooms;
```

Вывести номер (ID) первых 60 платежей, их суммы и даты оплаты.

```python
SELECT payment_id, amount_paid, payment_date
FROM payments
ORDER BY payment_id  -- Убедитесь, что вы сортируете по payment_id или другому столбцу, чтобы получить первые 60 записей
LIMIT 60;
```

Вывести всю информацию о каждом номере в порядке возрастания цены за ночь.

```python
SELECT room_number, type_name, price_per_night, room_size_sqm, max_occupancy, has_minibar, payment_option
FROM rooms
ORDER BY price_per_night ASC;
```

Вывести всю информацию о платежах в порядке убывания суммы платежа. Ограничьте вывод 60 записями.

```python
SELECT payment_id, booking_id, amount_paid, payment_date, payment_method
FROM payments
ORDER BY amount_paid DESC
LIMIT 60;
```
